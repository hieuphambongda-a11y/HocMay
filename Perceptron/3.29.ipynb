```python
import numpy as np

class Perceptron:

    def __init__(self, learning_rate=0.01, n_iters=10):
        self.learning_rate = learning_rate
        self.n_iters = n_iters
        self.w = None

    def fit(self, X, y):
        # Khởi tạo trọng số bằng 0
        self.w = np.zeros(X.shape[1])

        # Huấn luyện qua nhiều lần
        for i in range(self.n_iters):
            for j in range(len(X)):
                # Tính w^T.x
                z = np.dot(X[j], self.w)

                # Dự đoán nhãn
                if z >= 0:
                    y_pred = 1
                else:
                    y_pred = -1

                # Nếu dự đoán sai thì cập nhật w
                if y_pred != y[j]:
                    self.w = self.w + self.learning_rate * y[j] * X[j]

    def predict(self, X):
        # Tính X.w
        z = np.dot(X, self.w)
        # Dự đoán nhãn
        y_pred = np.where(z >= 0, 1, -1)
        return y_pred
```
