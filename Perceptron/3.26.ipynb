```python
import numpy as np
def f(x):
    return x**2 - 4*x + 5

def grad(x):
    return 2*x - 4

eta = 0.2
x = 5

print("Bước 0: x =", x, "f(x) =", f(x))
for i in range(4):
    g = grad(x)
    x = x - eta * g
    print("Bước", i + 1, ": x =", x, "f(x) =", f(x))
```

# Nhận xét:
- x giảm dần từ 5 xuống 2.3888.
- f(x) giảm từ 10 xuống 1.1511, x đang tiến gần đến 2
- thuật toán Gradient Descent đang hội tụ.
