```python
import numpy as np

# Khởi tạo w và x
w = np.array([-2, 1, 0])
x = np.array([2, 3, 1])

# Nhãn thực tế
y = 1

# Tính w^T x
z = np.dot(w, x)
print("w^T x ban đầu =", z)

# Xác định nhãn dự đoán
if z >= 0:
    y_pred = 1
else:
    y_pred = -1
print("Nhãn dự đoán =", y_pred)

# Kiểm tra phân lớp sai
if y_pred != y:
    print("Mẫu bị phân lớp sai")
    eta = 1
    # Cập nhật Perceptron
    w = w + eta * y * x
    print("w sau cập nhật =", w)

# Tính lại w^T x
z = np.dot(w, x)
print("w^T x sau cập nhật =", z)

# Kiểm tra lại nhãn dự đoán
if z >= 0:
    y_pred = 1
else:
    y_pred = -1
print("Nhãn dự đoán sau cập nhật =", y_pred)

# Kiểm tra đúng hay sai
if y_pred == y:
    print("Sau cập nhật: phân lớp đúng")
else:
    print("Sau cập nhật: phân lớp sai")
```
